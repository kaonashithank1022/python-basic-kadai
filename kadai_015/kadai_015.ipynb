```python
# 1. Humanクラスの定義
class Human:

    def __init__(self, name, age):
        # 属性（nameとage）の初期化
        self.name = name
        self.age = age

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print(f"名前: {self.name}")
        print(f"年齢: {self.age}")


# 2. Humanクラスのインスタンスを作成し、変数に代入する
human = Human("侍太郎", 20)

# 3. 定義したメソッドを実行して出力する
human.printinfo()
```
